```python
df = spark.table("samples.nyctaxi.trips")
display(df)
df.printSchema()
print("row count:", df.count())
```

```python
%sql
SELECT
  hour(tpep_pickup_datetime) AS pickup_hour,
  count(*)                   AS trips,
  round(avg(fare_amount), 2) AS avg_fare
FROM samples.nyctaxi.trips
GROUP BY hour(tpep_pickup_datetime)
ORDER BY pickup_hour
```

```python
from pyspark.sql import functions as F

# Step 1: add the derived column, and give the result a name
with_hour = df.withColumn("pickup_hour", F.hour("tpep_pickup_datetime"))
print(with_hour.columns)          # proof: pickup_hour is in the list now

# Step 2: aggregate — the linter can now see pickup_hour in with_hour's schema
result = (
    with_hour
    .groupBy("pickup_hour")
    .agg(
        F.count("*").alias("trips"),
        F.round(F.avg("fare_amount"), 2).alias("avg_fare"),
    )
    .orderBy("pickup_hour")
)
display(result)
```
